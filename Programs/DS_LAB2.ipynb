### Data Cleaning

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({ 'Age': [25,30,np.nan,40,35],
     'Department': ['HR','Finance','Finance',np.nan,'IT']})
print('Orginal Dataset(With Missing Values):')
print(df)
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({ 'Age': [25,30,np.nan,40,35],
     'Department': ['HR','Finance','Finance',np.nan,'IT']})
print('Orginal Dataset(With Missing Values):')
print(df)
df_ffill = df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)
df_bfill = df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({ 'Age': [25,30,np.nan,40,35],
     'Department': ['HR','Finance','Finance',np.nan,'IT']})
print('Orginal Dataset(With Missing Values):')
print(df)
df_drop = df.copy()
df_drop.dropna(inplace=True)
print(df_drop)
df_drop_col = df.copy()
df_drop_col.dropna(axis=1, inplace=True)
print(df_drop_col)
```

```python
import pandas as pd
df = pd.DataFrame({
    'ID': [1, 2,2, 3, 4, 5],
    'Name': ['John', 'Jane','Jane', 'Mike', 'Emily', 'David'],
    'Age': [25, 30,35,35, 28,32]
})
print('Original Dataset:')
print(df)
df_exact = df.drop_duplicates(subset = ['ID'])
print('Dataset after removing duplicates:')
print(df_exact)
```

```python
import pandas as pd
df = pd.DataFrame({
    'Date': ['2025-01-05', '05/01/2025', 'Jan 5,2025', '20250105', '2025.01.05'],
})
print('Original Dataset:')
print(df)
df['Date'] = pd.to_datetime(df['Date'],errors = 'coerce').dt.strftime('%y-%m-%d')
print(df)
```

```python
df = pd.DataFrame({
    'Name': ['John', 'Jane', 'Mike', 'Emily', 'David'],
})
print(df)
df['Name_lower'] = df['Name'].str.lower()
df['Name_upper'] = df['Name'].str.upper()
print("\n",df)
```

## Visualizations

#### Simple Plot

```python
import matplotlib.pyplot as plt
fig,ax = plt.subplots(figsize=(6,4))
ax.plot([1,2,3,4],[1,4,9,16])
ax.set_title('Simple Plot')
plt.show()
```

#### Basic Line Plot

```python
import matplotlib.pyplot as plt
x = [1,2,3,4]
y = [1,4,9,16]
plt.plot(x,y)
plt.show()
```

#### Multiple Line Plot

```python
import matplotlib.pyplot as plt
y1 = []
y2 = []
x = range(-100,100,10)
for i in x: y1.append(i**2)
for i in x: y2.append(-i**2)
plt.plot(x,y1,label='Square')
plt.plot(x,y2,label='Cube')
plt.ylim(-1000,1000)
plt.axhline(y=0,color='k')
plt.axvline(x=0,color='k')
plt.show()
```

#### Scatter Plot

```python
import matplotlib.pyplot as plt
x = [1,2,3,4]
y = [1,4,9,16]
plt.scatter(x,y,s=30,color='blue')
plt.show()
```

#### Scatter Plot with Labels

```python
import matplotlib.pyplot as plt
x = [1,2,3,4]
y = [1,4,9,16]
plt.scatter(x,y,s=30,color='blue',marker='o',label='Points')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.legend()
plt.show()
```

#### Vertical Bar Chart

```python
import matplotlib.pyplot as plt
x = [5,2,8,4]
names = ["A","B","C","D"]
plt.bar(names,x,color='blue')
plt.show()
```

#### Horizontal Bar Chart

```python
import matplotlib.pyplot as plt
x = [5,2,8,4]
names = ["A","B","C","D"]
plt.barh(names,x,color='yellow')
plt.show()
```

#### Bar Chart with Different Colors and Widths

```python
import matplotlib.pyplot as plt
x = [5,2,8,4]
names = ["One","Two","Three","Four"]
c1 = ['red','green','blue','yellow']
c2 = ['r','y']
plt.bar(names,x,color=c1,width = 0.5)
plt.bar(names,x,color=c2,width = -0.5)
plt.show()
```

#### Subplots

```python
import matplotlib.pyplot as plt
plt.subplot(1,2,1)
plt.plot([1,2,3,4],[1,4,9,16])
plt.subplot(2,2,2)
plt.plot([1,2,3,4],[1,4,9,16])
```
